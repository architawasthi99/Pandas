```python
#4 most revenue in each quater
temp_df = pd.concat([q1, q2, q3], keys=['Q1', 'Q2', 'Q3']).reset_index()
temp_df
```

```python
temp_df = pd.concat([q1, q2, q3], keys=['Q1', 'Q2', 'Q3']).reset_index()

temp_df['item_price'] = temp_df['item_price'].str.replace('$', '').str.strip().astype(float)

temp_df.groupby(['level_0', 'item_id'])['item_price'].sum().groupby(level=0).idxmax()
```
