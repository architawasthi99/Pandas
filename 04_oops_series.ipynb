```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


class marksanalysis:

  def __init__(self):
    self.marks=pd.Series(
        [85,22,58,99,100],
        index=['archit','arjun','ayushi','sundar','champaklal']
    )

  def display_marks(self):
    print(self.marks)

  def calculate_avg(self):
    print("Average Marks",self.marks.mean())

  def topper(self):
    print("Topper-",self.marks.idxmax())
    print("Highest Marks-",self.marks.max())

  def plot(self,kind='line'):
    self.marks.plot(kind=kind)
    plt.show()

obj=marksanalysis()

obj.display_marks()
print("  ")
obj.calculate_avg()
print(" ")
obj.topper()
print(" ")
topper_marks = obj.marks.max()
print(obj.marks[obj.marks == topper_marks])
print(" ")
obj.plot(kind="bar")
```
