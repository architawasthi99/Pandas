Q-5.Write a function to change the data type of given a column or a Series. Function takes series and data type as input, returns the converted series.

```python
def convert(new,dtype):
  new=new.astype(dtype)
  return new

new=pd.Series([22,33,44,55,66],dtype='int64')
result=convert(new,'int16')
print(result)
print(result.dtype)
```

Q-9
i. Read `items.csv` making `item_name` as index.
ii. Show no of nan values
ii. Item price is given in $, so convert it to rupees without currency symbol.
iii. Make data type of newly made series as float.
iv. Fill nan with mean of the series

```python
read=pd.read_csv('/content/items.csv',index_col='item_name').squeeze("columns")
print(read.size-read.count())
```

```python
read=read.str.replace('$','',regex=False).astype(float)
price_inr=read * 94
price_inr
```

```python
price_inr.dtype
```

```python
mean=price_inr.mean()

print(mean)

price_inr.fillna(mean)
```
