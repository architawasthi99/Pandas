```python
genres=movies.groupby('Genre')
genres.first()
```

```python
genres.nth(7)
```

```python
#get_group()
genres.get_group('Horror')
```

```python
genres.describe()
```

```python
genres.sample()
```

```python
#aggregate method
#passing dict

genres.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes':'sum',
        'Gross':'sum'
    }
)
```

```python
#passing list
genres[['Runtime', 'IMDB_Rating']].agg(['min', 'max', 'mean'])
```

```python
#looping in  groups
result = []
for group, data in genres:
    result.append(data[data['IMDB_Rating'] == data['IMDB_Rating'].max()])
df = pd.concat(result)
df
```

```python
# split (apply) combine
genres.apply(min)
```

```python
#find the number of each movies starting with A in each group
def foo(group):
  return group['Series_Title'].str.startswith('A').sum()
```

```python
genres.apply(foo)
```

```python
#ranking movies in each group
def movie_rank(group):
    group['gen_rank'] = group['IMDB_Rating'].rank(ascending=False)
    return group
```

```python
genres.apply(movie_rank)
```
