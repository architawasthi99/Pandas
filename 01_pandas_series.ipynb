**Series** **from** **List**

```python
import numpy as np
import pandas as pd
```

```python
#string
country=['india','usa','russia','china','aghanistan']
pd.Series(country)
```

```python
#integer
runs=[54,2,100,99,45]
pd.Series(runs)
```

```python
#custom index

marks=[10,20,30,40,50]
subjects=['hindi','english','physics','chemistry','history']
pd.Series(marks,index=subjects)
```

```python
#setting a name to series object
pd.Series(marks,index=subjects,name='archit marks')
```

**Series** **using** **dict**

```python
marks={
    'hindi':29,
    'english':100,
    'maths':99
}
new_marks=pd.Series(marks,name='archit hua pass')
new_marks
```
